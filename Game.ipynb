```python
# Cell 1 — Clone or pull latest from GitHub
import os,sys

REPO = "https://github.com/stanleyzihanfan/A_Game.git"
BRANCH = "feature"
PATH = "/content/repo"

if os.path.exists(PATH):
    # Repo already cloned this session — just pull latest
    !git -C {PATH} pull origin {BRANCH}
else:
    # Fresh clone
    !git clone -b {BRANCH} {REPO} {PATH}

if PATH not in sys.path:
    sys.path.insert(0, PATH)

# Remove any cached server/mod modules so running start server cell picks up file changes
mods_to_remove = [key for key in sys.modules if key.startswith("server") or key.startswith("mod_")]
for mod in mods_to_remove:
    del sys.modules[mod]

print("Repo ready at", PATH)
```
